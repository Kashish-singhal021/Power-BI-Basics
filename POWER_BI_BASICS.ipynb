#Power BI Basics


Question 1 : What is Power BI and why is it used in businesses?

-Power BI is a Business Intelligence (BI) and data visualization tool developed by Microsoft. It helps businesses collect, analyze, and visualize data from different sources and present it through interactive reports, charts, dashboards, and graphs.

Why is Power BI used in businesses?

Businesses use Power BI to:

Analyze data – Understand large amounts of business data easily.
Create dashboards and reports – Present information using charts, graphs, and tables.
Make better decisions – Help managers make data-driven decisions.
Track business performance – Monitor sales, profits, customers, expenses, and other KPIs.
Connect different data sources – Import data from Excel, databases, websites, and other sources.
Find trends and patterns – Identify changes and opportunities in business data.

Example: A retail company can use Power BI to create a dashboard showing monthly sales, profit, top-selling products, and sales by region, helping managers understand how the business is performing.

Question 2 : Name and explain the three main components of Power BI.

-The three main components of Power BI are:

Power BI Desktop
It is a desktop application used to connect to data, clean and transform data, create data models, and build interactive reports and dashboards.
Power BI Service
It is a cloud-based online platform where users can publish, share, view, and collaborate on Power BI reports and dashboards.
Power BI Mobile
It is a mobile application for Android and iOS that allows users to view and interact with reports and dashboards on smartphones and tablets.

Question 3 : Explain the Power BI workflow.

-The Power BI workflow is the process of taking raw data and converting it into useful reports and dashboards for decision-making.

Main steps in the Power BI workflow:
Connect to Data
First, Power BI connects to data sources such as Excel, CSV files, databases, websites, or cloud services.
Clean and Transform Data
Using Power Query, the data is cleaned by removing errors, handling missing values, changing data types, and transforming the data into a suitable format.
Create Data Model
Tables are connected using relationships, and calculations can be created using DAX (Data Analysis Expressions).
Create Visualizations
Charts, graphs, tables, cards, maps, and other visuals are created to understand the data.
Create Reports and Dashboards
Different visuals are arranged into interactive reports and dashboards.
Publish and Share
The report is published to Power BI Service, where it can be shared and accessed by other users.

Question 4 : List any four data cleaning tasks that can be performed in Power Query.

-Four common data cleaning tasks that can be performed in Power Query are:

Remove duplicates – Removes repeated or duplicate records from the data.
Handle missing values – Removes, replaces, or fills empty/null values.
Change data types – Changes columns to the correct type, such as text, number, date, or currency.
Remove errors – Identifies and removes or replaces incorrect/error values in the data.

Question 5 : Write step by step instructions to load the above dataset.

-To load the Real Estate Data CSV file into Power BI, follow these steps:

Open Power BI Desktop on your computer.
Click on Home → Get Data.
Select Text/CSV from the list of available data sources.
Browse and select the Real_Estate_Data_100_Rows.csv file.
Power BI will show a preview of the dataset. Check that the columns and data look correct.
Click Transform Data if you want to clean or modify the data using Power Query.
In Power Query, check the column names, data types, missing values, and duplicate records.

The dataset will now be loaded into Power BI Desktop, and you can use it to create charts, tables, reports, and dashboards.

Question 6 : Define Data View, Report View, and Model View. Explain the purpose of each view.

-| View            | Definition                                                       | Purpose                                                                                   |
| --------------- | ---------------------------------------------------------------- | ----------------------------------------------------------------------------------------- |
| **Data View**   | Shows the actual data stored in the tables.                      | Used to **view, inspect, and check data**, including rows, columns, and data types.       |

Question 8 : Split Owner Name to create two new columns as First Name and Last Name

-In Power Query, we can split the Owner Name column into two separate columns: First Name and Last Name.

Steps:
Open Power Query Editor by clicking Home → Transform Data.
Select the Owner Name column.
Go to the Transform tab.
Click Split Column → By Delimiter.
Select Space as the delimiter.
Choose Split at the left-most delimiter.
Click OK.
Power Query will create two columns.
Rename the first column to First Name.
Rename the second column to Last Name.
Click Home → Close & Apply to save the changes.
